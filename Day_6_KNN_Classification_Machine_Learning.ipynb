# Day 6: K-Nearest Neighbors (KNN) Classification

**Goal:** Classify data using the nearest examples.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 1. Create a dataset

```python
data = {
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8, 2, 3, 5, 6, 7, 4, 8, 1, 3, 6, 7, 5],
    "attendance": [45, 55, 60, 68, 75, 80, 90, 95, 50, 65, 78, 85, 88, 72, 92, 40, 58, 82, 93, 76],
    "result": ["Fail", "Fail", "Fail", "Fail", "Pass", "Pass", "Pass", "Pass",
               "Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass", "Fail",
               "Fail", "Pass", "Pass", "Pass"]
}
df = pd.DataFrame(data)
display(df)
print(df["result"].value_counts())
```

## 2. Select features (X) and label (y)

```python
X = df[["study_hours", "attendance"]]
y = df["result"]
print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 3. Split the data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 4. Scale the features

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 5. Train KNN

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled, y_train)
```

## 6. Predict and evaluate

```python
y_pred = knn.predict(X_test_scaled)
display(pd.DataFrame({"Actual": y_test, "Predicted": y_pred}))
print("Accuracy:", round(accuracy_score(y_test, y_pred), 3))
print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred, labels=knn.classes_))
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

## 7. Predict a new example

```python
new_student = pd.DataFrame({"study_hours": [5], "attendance": [78]})
new_scaled = scaler.transform(new_student)
print("Predicted result:", knn.predict(new_scaled)[0])
print("Class probabilities:", dict(zip(knn.classes_, knn.predict_proba(new_scaled)[0].round(3))))
```

## 8. Compare K values

```python
scores = []
for k in range(1, 8):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_scaled, y_train)
    pred = model.predict(X_test_scaled)
    scores.append({"k": k, "accuracy": accuracy_score(y_test, pred)})
display(pd.DataFrame(scores))
```

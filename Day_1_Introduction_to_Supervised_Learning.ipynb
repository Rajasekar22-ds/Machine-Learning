## 1. What is Machine Learning?

Machine Learning (ML) is a way for computers to learn patterns from data and use those patterns to make predictions.

## 2. What is Supervised Learning?

**Supervised learning is a type of machine learning in which a model learns from examples that include the correct answers.**

Imagine a teacher showing a student pictures and telling them the correct answer for each picture. After seeing many examples, the student tries to identify a new picture.

### Example: weld classification

| Example | Correct label |
|---|---|
| Photo of a weld with no visible issue | Good Weld |
| Photo of a weld with a defect | Bad Weld |
| Another labeled weld photo | Good Weld |

The model learns patterns from the examples and predicts a label for a new image.

*Note:* Real weld inspection requires suitable expert-labeled data and careful testing. A model prediction is not automatically a safety certification.

## 3. Three important terms

### Features (X)
Features are the input information used to make a prediction. Examples include flower measurements, image information, or customer purchase history.

### Label or target (y)
The label is the correct answer the model learns to predict. Examples include `Good Weld`, `Bad Weld`, a flower species, or a house's sale price.

### Training data
Training data is the set of examples used to teach the model. In supervised learning, examples include features and known targets.

Remember:
- **X = input features**
- **y = correct target/answer**

## 4. Two common supervised-learning tasks

We will learn these separately later. For now, remember the basic difference.

### Classification
Predicts a **category**.
Examples: Good Weld/Bad Weld, Spam/Not Spam, Cat/Dog.

### Regression
Predicts a **number**.
Examples: house price, temperature, estimated delivery time.

**Memory tip:** category → classification; number → regression.

## 5. A tiny Python example

This toy dataset shows features and known labels. It is only to understand the data format; it does not train an image-inspection model.

```python
import pandas as pd

data = pd.DataFrame({
    "weld_width_mm": [5.1, 4.8, 5.0, 3.2, 3.5],
    "surface_irregularity": [0.1, 0.2, 0.1, 0.8, 0.7],
    "label": ["Good Weld", "Good Weld", "Good Weld", "Bad Weld", "Bad Weld"]
})

display(data)
```

### Separate features (X) and labels (y)

The model receives `X` as input and learns to predict `y`.

```python
X = data[["weld_width_mm", "surface_irregularity"]]
y = data["label"]

print("Features (X):")
display(X)

print("Labels (y):")
display(y)
```

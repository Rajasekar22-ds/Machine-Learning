# Day 5: Logistic Regression

**Goal:** Predict a category using Logistic Regression.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## 1. Create a dataset

```python
data = {
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8, 2, 3, 5, 6, 7, 4, 8, 1],
    "attendance": [45, 55, 60, 68, 75, 80, 90, 95, 50, 65, 78, 85, 88, 72, 92, 40],
    "result": ["Fail", "Fail", "Fail", "Fail", "Pass", "Pass", "Pass", "Pass",
               "Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass", "Fail"]
}
df = pd.DataFrame(data)
display(df)
print(df["result"].value_counts())
```

## 2. Select X and y

```python
X = df[["study_hours", "attendance"]]
y = df["result"]
print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 3. Split the data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 4. Train Logistic Regression

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
print("Classes:", model.classes_)
```

## 5. Predict test results

```python
y_pred = model.predict(X_test)
display(pd.DataFrame({"Actual": y_test, "Predicted": y_pred}))
```

## 6. Evaluate

```python
print("Accuracy:", round(accuracy_score(y_test, y_pred), 3))
print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred, labels=model.classes_))
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

## 7. Predict a new example

```python
new_student = pd.DataFrame({"study_hours": [5], "attendance": [78]})
prediction = model.predict(new_student)
probabilities = model.predict_proba(new_student)
print("Predicted result:", prediction[0])
print("Class probabilities:", dict(zip(model.classes_, probabilities[0].round(3))))
```

## 8. Predict several examples

```python
new_students = pd.DataFrame({
    "study_hours": [1, 3, 5, 7],
    "attendance": [45, 60, 78, 90]
})
new_students["predicted_result"] = model.predict(new_students)
display(new_students)
```

# Day 2: Features (X) and Labels (y)

**Goal:** Separate input columns (`X`) and the target column (`y`) using pandas.

```python
import pandas as pd
```

## 1. Create a dataset

```python
data = {
    "weld_width": [5.2, 4.1, 6.0, 3.8, 5.5],
    "surface_irregularity": [1, 4, 0, 5, 2],
    "weld_quality": ["Good", "Bad", "Good", "Bad", "Good"]
}

df = pd.DataFrame(data)
df
```

## 2. Select features (X) and label (y)

```python
X = df[["weld_width", "surface_irregularity"]]
y = df["weld_quality"]

print("Features (X):")
display(X)

print("Labels (y):")
display(y)
```

## 3. Check shapes and columns

```python
print("Dataset shape:", df.shape)
print("Features shape:", X.shape)
print("Labels shape:", y.shape)

print("Feature columns:", X.columns.tolist())
print("Label name:", y.name)
```

## 4. Select one feature

```python
X_width = df[["weld_width"]]
X_irregularity = df[["surface_irregularity"]]

display(X_width)
display(X_irregularity)
```

## 5. Inspect the data

```python
print("First 3 rows:")
display(X.head(3))

print("Label counts:")
print(y.value_counts())
```

## 6. Another example

```python
students = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5],
    "attendance": [60, 65, 75, 85, 95],
    "result": ["Fail", "Fail", "Pass", "Pass", "Pass"]
})

X_students = students[["study_hours", "attendance"]]
y_students = students["result"]

display(students)
print("X:")
display(X_students)
print("y:")
display(y_students)
```

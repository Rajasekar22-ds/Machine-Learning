# Day 3: Classification in Machine Learning

**Goal:** Predict a category using a classification model.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## 1. Create a small dataset

```python
data = {
    "study_hours": [1, 2, 3, 4, 5, 6, 1, 2, 5, 6, 3, 4],
    "attendance": [50, 55, 65, 70, 85, 95, 45, 60, 90, 88, 72, 78],
    "result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass",
               "Fail", "Fail", "Pass", "Pass", "Pass", "Pass"]
}

df = pd.DataFrame(data)
display(df)
print(df["result"].value_counts())
```

## 2. Select features (X) and label (y)

```python
X = df[["study_hours", "attendance"]]
y = df["result"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 3. Split the data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 4. Train a classification model

```python
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
```

## 5. Make predictions

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})
display(results)
```

## 6. Evaluate the model

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred, labels=["Fail", "Pass"]))
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

## 7. Predict a new example

```python
new_student = pd.DataFrame({
    "study_hours": [4],
    "attendance": [80]
})

prediction = model.predict(new_student)
print("Predicted result:", prediction[0])
```

## 8. Try another model

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

knn_pred = knn.predict(X_test)
print("KNN accuracy:", accuracy_score(y_test, knn_pred))
```
